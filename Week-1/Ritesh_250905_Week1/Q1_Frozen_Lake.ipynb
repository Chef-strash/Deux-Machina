```python
import gymnasium as gym
import numpy as np
import random
```

```python

env =gym.make(
    'FrozenLake-v1',
    desc=None ,
    map_name="4x4",
    is_slippery=True,
    success_rate=1.0/3.0,
    reward_schedule=(100, -50, -1)
)

states= env.observation_space.n
actions= env.action_space.n
```

```python
alpha =0.8
gamma = 0.95
episodes =20000
steps=100
eps_start=1.0
eps_end=0.01
eps_decay=0.99
Q = np.zeros((states, actions))
```

```python
for episode in range(episodes):
   S,_ =env.reset()
   epsilon = max(eps_end,eps_start*(eps_decay**episode))

   for step in range(steps):
    if random.uniform(0,1) < epsilon :
      A= env.action_space.sample()
    else :
      A=np.argmax(Q[S])
    s , R , terminated , truncated ,_= env.step(A)
    q=np.max(Q[s])
    Q[S,A]=Q[S,A] +alpha*(R+gamma*q-Q[S,A])

    S=s

    if terminated or truncated :
      break
print("training done")  #checking if this block works or not
np.save("Qtable.npy",Q)
Q_loaded=np.load("Qtable.npy")
print(np.max(Q))
print(np.min(Q))
print(Q[0])
#checking if the Q values are actually being changed or not
```

```python
#using the training done in cell 4 for testing below

test_ep= 100
for ep in range(test_ep):
    S,_ = env.reset()
    print(f"\nEpisode {ep+1}")

    for step in range(steps):
        A = np.argmax(Q_loaded[S])
        S, R, terminated, truncated,_ = env.step(A)

        if terminated or truncated:
            if R > 0:
                print("Reached goal")
            else:
                print("Fell in the hole")
            break
```
