```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from google.colab import files
files.upload()
import statistics
from sklearn.model_selection import train_test_split
import statsmodels.api as sm

```

```python
sal_yrs = pd.read_csv("extended_salary_data.csv")
sal_yrs
```

```python

plt.scatter(sal_yrs['YearsExperience'], sal_yrs['Salary'])
plt.xlabel = 'years_of_experience'
plt.ylabel = 'salary'
plt.show
```

```python
avg_yrs = statistics.mean(sal_yrs['YearsExperience'])
median_yrs = statistics.median(sal_yrs['YearsExperience'])
variance_yrs = statistics.variance(sal_yrs['YearsExperience'])
avg_sal = statistics.mean(sal_yrs['Salary'])
median_sal = statistics.median(sal_yrs['Salary'])
variance_sal = statistics.variance(sal_yrs['Salary'])
print('mean of yrs = ', avg_yrs)
print('median of yrs = ', median_yrs)
print('variance of yrs = ', variance_yrs)
print('mean of salary = ', avg_sal)
print('median of salary = ', median_sal)
print('variance of salary = ', variance_sal)
```

```python
x = sal_yrs['YearsExperience']
y= sal_yrs['Salary']
x_train, x_test, y_train, y_test = train_test_split( x, y, train_size = 0.7, test_size = 0.3, random_state = 100 )
```

```python
# Add a constant to get an intercept
x_train_sm = sm.add_constant(x_train)
# Fit the resgression line using 'OLS'
reg_line = sm.OLS(y_train, x_train_sm).fit()
```

```python
reg_line.params
```

```python
x_test_sm = sm.add_constant(x_test)
y_pred = reg_line.predict(x_test_sm)
y_pred.head()
```

```python
#Imporitng libraries
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
#RMSE value
print( "RMSE: ",np.sqrt( mean_squared_error( y_test, y_pred ) ))
#R-squared value
print( "R-squared: ",r2_score( y_test, y_pred ) )
```
