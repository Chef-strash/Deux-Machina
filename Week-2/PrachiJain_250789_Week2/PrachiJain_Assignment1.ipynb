```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
%matplotlib inline
```

```python
df= pd.read_csv("../Dataset/extended_salary_data.csv")
plt.scatter(df.iloc[:, 0], df.iloc[:, 1])
plt.xlabel("Years of Experience (in yrs)")
plt.ylabel("Annual Salary (in USD)")
plt.title("Data of Employees")
```

```python
for i in [0, 1]:
    if(i==0): print("Metrics of Years of Experience (in yrs): ")
    else : print("Metrics of Annual Salary (in USD): ")
    print("Count= ", df.iloc[:, i].count())
    print("Mean= ", df.iloc[:, i].mean())
    print("Median= ", df.iloc[:, i].median())
    print("Mode= ", df.iloc[:, i].mode().iloc[0])
    print("Standard Deviation= ", df.iloc[:, i].std())
    print("Variance= ", df.iloc[:, i].var())
    print("Max= ", df.iloc[:, i].max())
    print("Min= ", df.iloc[:, i].min())
    print()
```

```python
X = df['YearsExperience'].values.reshape(-1, 1)
Y = df['Salary'].values
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
plt.scatter(X_train, Y_train, color='blue', label='Training_Data')
plt.plot(X_train, model.predict(X_train), color='red', label='Best_Fit_Line')
plt.xlabel("Years of Experience (in yrs)")
plt.ylabel("Annual Salary (in USD)")
plt.title("Training Result")
plt.legend()
```

```python
Y_pred = model.predict(X_test)
plt.scatter(X_test, Y_test, color='blue', label='Testing_Data')
plt.plot(X_test, Y_pred, color='red', label='Best_Fit_Line')
plt.xlabel("Years of Experience (in yrs)")
plt.ylabel("Annual Salary (in USD)")
plt.title("Testing Result")
plt.legend()
```

```python
print("Evalutaion Metrics-")
print("Mean Absolute Error (MAE): ", mean_absolute_error(Y_test, Y_pred))
print("Mean Squared Error (MSE): ", mean_squared_error(Y_test, Y_pred))
print("Root Mean Squared Error (RMSE): ", np.sqrt(mean_squared_error(Y_test, Y_pred)))
print("R2 Score: ", r2_score(Y_test, Y_pred))
```

```python
print("Results-")
print("Slope: ", model.coef_[0])
print("Intercept: ", model.intercept_)
```
