```python
import pandas as pd
import numpy as np
url="https://raw.githubusercontent.com/Chef-strash/Deux-Machina/refs/heads/main/Week-2/Dataset/extended_salary_data.csv"
df=pd.read_csv(url)
df.head()
```

```python
import matplotlib.pyplot as plt
x=df["YearsExperience"]
y=df["Salary"]
plt.scatter(x, y)
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Scatter Plot")
plt.show()
print("Mean:")
print(df.mean().round(2))
print("")
intervals = [0, 5, 10, 15, 20]   # choose class intervals
groups = pd.cut(x, intervals)
print("Mode:")
print(groups.max())
print("")
print("Median:")
print(df.median().round(2))
print("")
print("Variance:")
print(df.var().round(1))
print("")
print("Std:")
print(df.std().round(2))
print("")
```

```python
#We can see from above plot that it is linearly varying
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=4)

x_train=x_train.values.reshape(-1,1)
x_test=x_test.values.reshape(-1,1)
y_train=y_train.values.reshape(-1,1)
y_test=y_test.values.reshape(-1,1)

from sklearn.linear_model import LinearRegression
Linear_model = LinearRegression()
Linear_model.fit(x_train, y_train)
```

```python
y_pred=Linear_model.predict(x_test)
from sklearn import metrics
slope=Linear_model.coef_
intercept=Linear_model.intercept_
print("Slope:",slope.round(2))
print("Intercept:",intercept.round(2))
print("Mean Squared Error:",round(metrics.mean_squared_error(y_test,y_pred),2))
print("Mean Absolute Error:",round(metrics.mean_absolute_error(y_test,y_pred),2))
```
